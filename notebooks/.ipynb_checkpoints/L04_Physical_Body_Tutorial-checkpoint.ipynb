## Part 2: 2D Kinematics, Jacobian Velocity Control, and Singularity Resolution (E5–E8) (1 Pt = 0.25 Pt × 4)

**Title:** Aligning the Claw Parallel Gripper with the Milk Carton

![Alternative Text](tutorial4_2d_mobile_manipulator.png)

### Scenario & System Parameters (Refer to Schematic Diagram)
Consider a planar 2D mobile manipulator operating in the $(x, y)$ vertical plane in front of a table:
* **Mobile Base:** Rectangular chassis ($W = 0.50\text{ m}$, $H = 0.25\text{ m}$, mass $m_{\text{base}} = 25\text{ kg}$) mounted on wheels of radius $r_w = 0.10\text{ m}$. Initially, the base center is at $x_{\text{base}}(0) = 0.30\text{ m}$ on the floor ($y = 0$).
* **Shoulder Pivot (Joint 1):** Mounted on the top-front of the chassis at offset $(+0.15\text{ m}, +0.45\text{ m})$ relative to $(x_{\text{base}}, 0)$, giving initial world coordinates $(x_s, y_s) = (0.45\text{ m}, 0.45\text{ m})$.
* **3-DOF Planar Arm + Claw Parallel Gripper:**
  * Link 1 (Upper arm): length $L_1 = 0.45\text{ m}$, mass $m_1 = 3.0\text{ kg}$, joint angle $q_1$ (measured counter-clockwise from $+x$).
  * Link 2 (Forearm): length $L_2 = 0.40\text{ m}$, mass $m_2 = 2.0\text{ kg}$, relative joint angle $q_2$.
  * Link 3 (Claw parallel gripper): length $L_g = 0.12\text{ m}$, mass $m_g = 0.5\text{ kg}$, relative wrist angle $q_3$.
  * Total gripper orientation in world frame: $\phi_g = q_1 + q_2 + q_3$.
* **Table & Target Object:**
  * Table surface height: $H_{\text{table}} = 0.72\text{ m}$ (left edge at $x_{\text{table}} = 1.25\text{ m}$).
  * Milk carton resting upright on the table: width $w_m = 0.08\text{ m}$, height $h_m = 0.20\text{ m}$, mass $m = 1.06\text{ kg}$, center at $(x_{\text{milk}}, y_{\text{milk}}) = (1.48\text{ m}, 0.82\text{ m})$.

---

1. **Forward Kinematics (0.25 Pt):**
   Write the forward kinematics equations expressing the Tool Center Point (TCP) pose $[x_{\text{tcp}}, y_{\text{tcp}}, \phi_g]^T$ as a function of the base position $x_{\text{base}}$ and joint angles $q = [q_1, q_2, q_3]^T$.

2. **Inverse Kinematics for Horizontal Grasp Alignment (0.25 Pt):**
   To grasp the milk carton cleanly from the left without colliding with the table, the gripper must approach horizontally ($\phi_g = 0^\circ$) with its TCP reaching the carton's center $(x_{\text{tcp}}^*, y_{\text{tcp}}^*) = (1.48\text{ m}, 0.82\text{ m})$.
   * With the base fixed at $x_{\text{base}}(0) = 0.30\text{ m}$, first determine whether the target is reachable by computing the required distance from the shoulder $(x_s, y_s)$ to the wrist $(x_w^*, y_w^*) = (x_{\text{tcp}}^* - L_g, y_{\text{tcp}}^*)$.
   * If the base moves to $x_{\text{base}} = 0.55\text{ m}$ (placing the shoulder at $(0.70\text{ m}, 0.45\text{ m})$), compute the **elbow-up** ($q_2 < 0$) inverse kinematics solution $(q_1, q_2, q_3)$ that aligns the gripper horizontally at the milk carton center.

3. **Jacobian Matrix & Joint Velocity Control (0.25 Pt):**
   * Derive the $3 \times 3$ analytical Jacobian $J(q) = \frac{\partial [x_{\text{tcp}}, y_{\text{tcp}}, \phi_g]^T}{\partial [q_1, q_2, q_3]^T}$ for the fixed-base arm.
   * During the final pre-grasp approach, the gripper must move horizontally toward the milk carton at Cartesian velocity $\dot{x} = [v_x, v_y, \omega]^T = [0.15\text{ m/s}, 0.0\text{ m/s}, 0.0\text{ rad/s}]^T$. Compute the required joint velocities $\dot{q} = J(q)^{-1}\dot{x}$ at the configuration found in E2.

4. **Singularity Analysis & Base Wheel Compensation (0.25 Pt):**
   * Suppose the milk carton is placed deeper on the table at $(x_{\text{milk}}, y_{\text{milk}}) = (1.67\text{ m}, 0.45\text{ m})$ while the base is at $x_{\text{base}} = 0.30\text{ m}$. Show that reaching the target forces the arm into a **boundary kinematic singularity** ($q_2 = 0^\circ$, $q_3 = 0^\circ$, $\det(J) = 0$).
   * Explain physically what happens to the manipulator's velocity capability along the $+x$ direction at this singularity.
   * How far must the mobile base translate ($\Delta x_{\text{base}}$), and by what angle $\Delta \phi_w$ (in radians and degrees) must the wheels of radius $r_w = 0.10\text{ m}$ rotate (assuming rolling without slipping, $\Delta x_{\text{base}} = r_w \Delta \phi_w$), so that the arm can reach the carton comfortably with an elbow bend of $q_2 = -45^\circ$ and $\phi_g = 0^\circ$?

## Part 3: Simplified Planar Dynamics Model (E9–E11) (1 Pt)

**Title:** Motor Torques and Actuation Forces Neglecting Velocity-Dependent Nonlinearities

In general, rigid-body manipulator dynamics follow $M(q)\ddot{q} + C(q, \dot{q})\dot{q} + B(\dot{q}) + g(q) = \tau$. Here, we neglect air resistance, viscous friction ($B(\dot{q}) \approx 0$), and centrifugal/Coriolis forces ($C(q, \dot{q})\dot{q} \approx 0$), retaining only **link/base masses (inertia)**, **gravity** ($g = 9.81\text{ m/s}^2$), and **motor torques/forces**:

$$M(q)\ddot{q} + g(q) = \tau$$

---

1. **Simplified Equation of Motion for the Mobile Base (0.30 Pt):**
   * Consider the mobile base moving horizontally along the floor ($y = \text{const}$, so gravity does no work along $x_{\text{base}}$). Let $m_{\text{total}} = m_{\text{base}} + m_1 + m_2 + m_g = 30.5\text{ kg}$ be the total translated mass of the robot.
   * Write the linear equation of motion relating the net horizontal traction force $F_{\text{wheel}}$ to base acceleration $\ddot{x}_{\text{base}}$, and compute the total wheel motor torque $\tau_w = F_{\text{wheel}} \cdot r_w$ required to accelerate the robot from rest at $\ddot{x}_{\text{base}} = 0.8\text{ m/s}^2$.

2. **Point-Mass Approximation of Arm Joint Torques (0.40 Pt):**
   * Model the mass of each arm link as concentrated at its distal end ($m_1 = 3.0\text{ kg}$ at distance $L_1$, $m_2 = 2.0\text{ kg}$ at distance $L_2$, and $m_g = 0.5\text{ kg}$ at distance $L_g$).
   * At the horizontal stretched posture $q = [0^\circ, 0^\circ, 0^\circ]^T$ with zero angular acceleration ($\ddot{q} = 0$), compute the **static gravity-holding torques** $g(q) = [\tau_{g,1}, \tau_{g,2}, \tau_{g,3}]^T$ required at the shoulder, elbow, and wrist motors.

3. **Dynamic Torque Under Joint Acceleration (0.30 Pt):**
   * Suppose the arm starts from the horizontal posture $q = [0^\circ, 0^\circ, 0^\circ]^T$ and initiates an upward lift with elbow angular acceleration $\ddot{q}_2 = 1.5\text{ rad/s}^2$ (while $\ddot{q}_1 = 0$, $\ddot{q}_3 = 0$). Using the diagonal inertia approximation $I_{22} \approx m_2 L_2^2 + m_g (L_2 + L_g)^2$, compute the total motor torque $\tau_2 = I_{22}\ddot{q}_2 + \tau_{g,2}$ required at the elbow motor.

## Part 4: Grasping, Contact Force Transmission, and Object Compliance (E12–E15) (1 Pt = 0.25 Pt × 4)

**Title:** Lifting the Milk Carton Without Slip or Structural Crushing

Once the claw parallel gripper is aligned around the milk carton ($m = 1.06\text{ kg}$, width $w_m = 0.08\text{ m}$, static friction coefficient $\mu_s = 0.35$, wall stiffness $k_{\text{obj}} = 800\text{ N/m}$), the fingers close symmetrically to establish a pinch grasp and lift the carton vertically off the table.

---

1. **Contact Force Transmission & Slip Edge (0.25 Pt):**
   * Draw/describe the free-body diagram of the milk carton during a vertical lift with upward acceleration $a_y = 2.0\text{ m/s}^2$ ($g = 9.81\text{ m/s}^2$).
   * Using the Coulomb friction law across both parallel finger pads ($2 \mu_s F_n \ge m(g + a_y)$), compute the minimum normal grasp force $F_{n,\min}$ required per finger so the carton does not slip.

2. **Object Stiffness & Allowable Gripper Deformation (0.25 Pt):**
   * Each side wall of the milk carton behaves as a linear elastic spring with stiffness $k_{\text{obj}} = 800\text{ N/m}$, such that $F_n = k_{\text{obj}} \Delta d$, where $\Delta d$ is the inward indentation per side.
   * If structural denting/rupture occurs when indentation exceeds $\Delta d_{\max} = 0.030\text{ m}$ ($30\text{ mm}$), determine the **admissible grasp force corridor** $[F_{n,\min}, F_{n,\max}]$ and the corresponding finger closure width window $w_{\text{grip}} = w_m - 2\Delta d$.

3. **Effect of Surface Moisture (Wet Carton Disturbance) (0.25 Pt):**
   * Condensation reduces the static friction coefficient from $\mu_s = 0.35$ (dry) to $\mu_{s,\text{wet}} = 0.18$ (wet). Recompute $F_{n,\min}$ for the same lift acceleration $a_y = 2.0\text{ m/s}^2$. Can the robot safely lift the wet carton without exceeding the denting threshold $\Delta d_{\max} = 30\text{ mm}$?

4. **Static Wrench Transmission to Arm Motors (0.25 Pt):**
   * While holding the lifted milk carton stationary ($a_y = 0$) at the grasp configuration from Part 2 E2, the carton's weight exerts a downward external force $f_{\text{ext}} = [0, -mg, 0]^T$ on the TCP.
   * Using the Jacobian transpose relation $\tau_{\text{hold}} = g(q) - J(q)^T f_{\text{ext}}$, explain how the external payload force propagates back into the shoulder, elbow, and wrist motor torques, and compute the additional torque $\Delta \tau = -J(q)^T f_{\text{ext}}$ induced by the milk carton.
