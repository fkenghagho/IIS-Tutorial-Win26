**Title:** The Body Motion Problem: Serving Eva a Drink & Breakfast

### Scenario Background
Eva says: "Serve me a drink."

The kitchen contains two milk cartons in the fridge door:
- **Carton A:** Opened whole milk.
- **Carton B:** Unopened oat drink.

Eva's usual mug already contains hot black coffee on the counter.

A wet surface on Carton A causes slip during a low test lift, requiring a grasp adaptation before pouring.

2. **E2 · Constrained Action Designator & Grounding**

## Part 1: Solutions

---

#### E1 · The Gap & Whole-Course Judgment
* **Why nominal end-state success fails:** Achieving a final pose (e.g., mug on the table) does not solve the Body Motion Problem if unwanted side effects occur along the execution path. Task success requires satisfying constraints across the **entire trajectory**, not only at termination.
* **Two violated whole-course constraints:**
  1. *No spill / containment preserved:* Milk spilled across the counter during transport or pouring.
  2. *Environment / state restored safely:* The fridge door was left open instead of being closed after retrieval.

#### E2 · Constrained Action Designator & Grounding
* **Working Interpretation:** `(:type serving) (:recipient Eva) (:substance (some substance (:type coffee) (with milk)))` *(or `milk-for-coffee`)*
* **Inferred Function/Means:** `(:vessel-role contain-and-support)` *(inferring a vessel that contains and supports the drink for convenient serving)*
* **Grounded Physical Object:** `(:object carton-A)` *(grounded to the opened whole-milk carton on the fridge door shelf, distinct from Carton B oat drink and Eva's mug already holding black coffee)*

#### E3 · The Commitment Decision & Asymmetric Failure Costs
* **Predicted effect:** A **5 mm test lift near the shelf** checks whether the side pinch holds or slips while the carton is still right above its support surface. By contrast, a **high carry directly to the counter** lifts the carton high and moves it across the kitchen before grip reliability is verified.
* **Cost of failure:** Slipping during a **5 mm lift near the shelf** is *low-cost and recoverable* (the carton sets back down within millimeters with little or no spill); slipping during a **high carry** causes a *high-cost drop and spill*.
* **Decisive evidence:** The carton's **weight/fill level and surface wetness/friction are uncertain** prior to lifting, so a low test lift is warranted to observe actual grip stability before committing to a high carry.

#### E4 · Causal Prediction Across Material Variants
* **Stopping rules (`(:stop-when ...)`):**
  * *Milk into coffee:* `(:stop-when (>= poured-volume 26.5 mL))` *(stops early before the ~30 mL target because liquid in the spout and sensor/control lag continue arriving after the command)*
  * *Pancake batter into a pan:* `(:stop-when (>= disc-diameter 9 cm))` *(stops early before the ~12 cm target disc diameter because thick batter continues spreading outward after pouring stops)*
* **How material dynamics change wrist tilt-back:** Low-viscosity milk stops flowing quickly once tilted back, whereas viscous batter has a slower flow/cut-off and continues spreading on the pan after tilt-back—meaning tilt-back must be initiated **earlier** relative to the final target diameter.

#### E5 · System Decomposition & The Reasoning Loop
* **Triggered step:** **Detect & Revise** (paired with **Belief Update** and **Regulate**). When downward slip is detected during the 5 mm lift, the robot updates its belief that a side pinch is insufficient on the wet carton.
* **Action designator development:** The designator is revised from a side-pinch lift (`(:grasp side-pinch)`) to a **bottom-support grasp** (`(:grasp (:type bottom-support) (:hand palm-under))`) to carry the vertical load safely, and later refined to a **bimanual grasp** (e.g., left palm-under + right side-pinch) when preparing to tilt for pouring.

#### E6 · Conceptual Ablation & Evidence of Competence
* **Failure variant:** In a kitchen where **Carton A is wet/slippery or heavier than expected**, a fixed default high-side-pinch lift without a check-before-commitment skips testing near the shelf and can slip/drop milk mid-carry.
* **Single-run success vs. evidence of competence:** A single successful run, even a "lucky success" with a fixed default only shows that one execution worked under one set of conditions. **Evidence of general competence** requires adapting across task and environment variants (different wetness, weights, containers, and pour/handover constraints).

3. URDF of the robot base
